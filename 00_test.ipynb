```python
#If elif else statement for marks of students
#Above 94-Grade A, 90-94-Grade B, 85-89-Grade C, 75-84-Grade D, Below 75-Grade F

marks = int(input("Enter your marks: "))

if marks > 94:
    print("Grade A")
elif marks >= 90 and marks <= 94:
    print("Grade B")
elif marks >= 85 and marks <= 89:
    print("Grade C")
elif marks >= 75 and marks <= 84:
    print("Grade D")
else:
    print("Grade F")
```
